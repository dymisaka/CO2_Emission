```python
import os

file_list = []
for root, dirs, files in os.walk('.'):
    for file in files:
        if file.endswith('.xlsx'):
            file_list.append(os.path.join(root, file))
print(file_list)
file_num=len(file_list)
print("The number of files=",file_num)
```

```python
import pandas as pd


for i in range(file_num):
    if i==0:
        excel_file = pd.ExcelFile(file_list[i])
        sheet_names = excel_file.sheet_names
        data = excel_file.parse('Sheet1')
        
        # 统一col name
        rename_list = [col for col in data.columns if col.endswith('_df1')]
        data.columns = [col.replace('_df1', '') if col.endswith('_df1') else col for col in data.columns]

        print(file_list[i]+'的数据基本信息：')
        data.info()
#         print(data.head())
        data.insert(loc=1, column='type_no', value=i)
        continue
        
        
    # read xlsx file
    excel_file = pd.ExcelFile(file_list[i])
    sheet_names = excel_file.sheet_names
    df = excel_file.parse('Sheet1')
    
    # 统一col name
    rename_list = [col for col in df.columns if col.endswith('_df1')]
    df.columns = [col.replace('_df1', '') if col.endswith('_df1') else col for col in df.columns]
    print(file_list[i]+'的数据基本信息：')
    df.info()
#     print(df.head())
    df.insert(loc=1, column='type_no', value=i)
    data=pd.merge(data, df, how='outer')

data.info()
data.to_csv('./Updated Data/merged_data.csv')
```

```python
data = pd.read_csv('./Updated Data/power_data.csv')
# 处理 draught 列中的复合值数据
data['draught'] = data['draught'].str.split(' ').str[0].str.replace(',', '')

# 处理 engine_power 列中的复合值数据
data['engine_power'] = data['engine_power'].apply(lambda x: str(x).replace(',', '') if pd.notnull(x) else x)

# 使用 to_numeric 函数统一处理 datraught 列和 engine_power 的数据格式
data['draught'] = pd.to_numeric(data['draught'], errors='coerce')
data['engine_power'] = pd.to_numeric(data['engine_power'], errors='coerce')
# 删除空值行
data = data.dropna()
data.info()
# 保存文件
csv_path = './Updated Data/power_data_handle.csv'
data.to_csv(csv_path)
```
